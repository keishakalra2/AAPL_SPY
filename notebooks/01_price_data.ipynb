# Step 2 — Download AAPL and SPY price data

This notebook downloads daily price data for Apple (`AAPL`) and the S&P 500 ETF (`SPY`) and saves an untouched copy in `data/raw/`. Momentum and prediction targets will be added in a later step.

```python
from pathlib import Path

import pandas as pd
import yfinance as yf

TICKERS = ["AAPL", "SPY"]
START_DATE = "2023-01-01"
# yfinance treats end as exclusive. Jan. 2 includes the requested Jan. 1 cutoff.
DOWNLOAD_END = "2025-01-02"

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)
```

```python
prices = yf.download(
    TICKERS,
    start=START_DATE,
    end=DOWNLOAD_END,
    auto_adjust=False,
    progress=False,
)

if prices.empty:
    raise RuntimeError("No price data was downloaded. Check the internet connection and try again.")

prices.head()
```

```python
raw_path = RAW_DATA_DIR / "aapl_spy_2023-01-01_to_2025-01-01.csv"
prices.to_csv(raw_path)
print(f"Saved {len(prices):,} trading days to {raw_path}")
```
